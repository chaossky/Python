```python
from sympy import Matrix, symbols, solve, latex, eye
from IPython.display import display, Math

λ = symbols('λ')

def show_eigenvalues_latex(mat):
    # 단위 행렬 생성
    mat_I = eye(mat.shape[0])
    # 특성다항식
    char_poly = (mat - λ*mat_I).det()
    # 근(고유값)
    roots = solve(char_poly, λ)
    # LaTeX로 출력
    display(Math(" , ".join([latex(r) for r in roots])))

# 사용 예시
mat_test= Matrix([[-4, 1], [1, 3]])
show_eigenvalues_latex(mat_test)
```

```python
from sympy import Matrix, symbols, eye, roots, latex
from IPython.display import display, Math

λ = symbols('λ')

def show_eigenvalues_latex(mat):
    mat_I = eye(mat.shape[0])
    char_poly = (mat - λ*mat_I).det()
    # roots() 함수 사용 → 중복근까지 포함
    eigenvalues = roots(char_poly, λ, multiple=True)
    display(Math(" , ".join([latex(r) for r in eigenvalues])))

# 테스트
mat_test = Matrix([[3,0,0],[-2,4,0],[-4,2,3]])
show_eigenvalues_latex(mat_test)
```

### 대칭 행렬의 고유값 분해 특징

#### 📘 정의
   - 대칭 행렬(Symmetric matrix) (즉, $ A^T = A $)은 항상 특별한 성질을 가지며, 
   - 고유값 분해(Eigendecomposition)에서 다음과 같은 두 가지 핵심 특징을 갖습니다.
---

#### 🔑 특징
1. **실수 고유값**
   - 대칭 행렬은 항상 실수 고유값을 가집니다.
   - 이는 대칭 행렬이 실수 에르미트(Hermitian) 행렬의 특수한 경우이기 때문입니다.
   - 에르미트 행렬의 고유값은 반드시 실수입니다.

2. **직교하는 고유벡터**
   - 서로 다른 고유값에 대응하는 고유벡터들은 서로 직교합니다.
   - 따라서 대칭 행렬은 직교 행렬(orthogonal matrix)에 의해 대각화가 가능합니다.
   - 즉, 다음과 같이 분해할 수 있습니다:

   $$
   A = Q \Lambda Q^T
   $$

   - $A$ 는 행렬, $Q$는 직교행렬, $\Lambda$ 는 대각행렬 
---

#### 📊 요약
- 대칭 행렬의 고유값은 **항상 실수**이다.  
- 대칭 행렬의 고유벡터는 **서로 직교**한다.  
- 따라서 대칭 행렬은 **직교 대각화(orthogonal diagonalization)** 가 가능하다.
- 즉, 두 고유 벡터의 dot product 값은 0 입니다.


# Hermitian 행렬 (에르미트 행렬)

## 📘 정의
Hermitian 행렬은 복소수 정방행렬 중에서 자기 자신과 켤레 전치(conjugate transpose)가 같은 행렬을 말합니다.  
즉,
$$
A = A^* = \overline{A}^T
$$
를 만족하는 행렬이 Hermitian 행렬입니다.  
여기서 $\overline{A}$는 각 원소의 복소수 켤레, $T$는 전치(transpose)입니다.

- **실수 행렬의 경우**: 모든 원소가 실수라면 켤레가 필요 없으므로, Hermitian 조건은 단순히 **대칭 행렬** 조건과 동일합니다.  
  따라서 실수 대칭 행렬은 Hermitian 행렬의 특수한 경우입니다.

---

## 🔑 성질
1. **대각 원소는 항상 실수**  
   Hermitian 행렬의 대각 성분은 반드시 실수입니다.

2. **고유값은 항상 실수**  
   Hermitian 행렬의 모든 고유값은 실수입니다.

3. **고유벡터는 직교**  
   서로 다른 고유값에 대응하는 고유벡터들은 서로 직교합니다.

4. **직교 대각화 가능**  
   Hermitian 행렬은 항상 직교(또는 유니타리) 행렬에 의해 대각화할 수 있습니다.

---

## 📊 예시
- **실수 대칭 행렬**:
  $$
  \begin{bmatrix}
  2 & 1 \\
  1 & 2
  \end{bmatrix}
  $$
  → Hermitian 행렬 (실수 대칭이므로).

- **복소수 Hermitian 행렬**:
  $$
  \begin{bmatrix}
  2 & i \\
  -i & 3
  \end{bmatrix}
  $$
  → 전치 후 켤레를 취하면 원래 행렬과 같으므로 Hermitian.

---

## ✅ 요약
- Hermitian 행렬은 **복소수 대칭 행렬의 일반화**로, $A = A^*$를 만족합니다.  
- **고유값은 실수**, **고유벡터는 직교**라는 중요한 성질을 가지며, 이는 선형대수학과 데이터 과학(예: PCA, 양자역학, 신호처리)에서 핵심적으로 활용됩니다.


```python
import sympy as sp

# 고유벡터 행렬 P
P = sp.Matrix([[2, 3],
               [-4, 2]])

# 고유값 대각행렬 D
D = sp.Matrix([[1, 0],
               [0, 3]])

# 원래 행렬 A = P * D * P^{-1}
A = P * D * P.inv()

print("원래 행렬 A:")
sp.pprint(A)
```
