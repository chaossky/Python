```python
import py3Dmol, random
V=py3Dmol.view(width=400, height=400)
V.addSphere({"center":{"x":0,"y":0,"z":0},
             "radius":1.2,"color":"red"})

for _ in range(80):
    V.addSphere({
        "center":{k:random.uniform(-4,4) for k in "xyz"},
        "radius":0.2,
        "color":"cyan",
        "opacity":0.5
    })

V.setBackgroundColor("black")
V.zoomTo()
display(V.show())
```
