```python
from sympy import Matrix, symbols, solve, latex, eye
from IPython.display import display, Math
λ = symbols('λ') 
mat_A=Matrix([[-4,1],[1,3]])
mat_I=eye(2)

sik1=mat_A-λ*mat_I
sik2=sik1.det()
roots=solve(sik2,λ)
display(Math(" , ".join([latex(r) for r in roots])))
```

```python
import sympy as sp

# 행렬 정의
A = sp.Matrix([[1, 2],
               [3, -4]])

# 대각화 수행
P, D = A.diagonalize()

print("원래 행렬 A:")
sp.pprint(A)

print("\n고유벡터 행렬 P:")
sp.pprint(P)

print("\n대각행렬 D:")
sp.pprint(D)

# 확인: A = P * D * P^-1
A_reconstructed = P * D * P.inv()
print("\nP * D * P^-1 결과:")
sp.pprint(A_reconstructed)

# 역행렬 확인: A^-1 = P * D^-1 * P^-1
D_inv = D.inv()
A_inv = P * D_inv * P.inv()
print("\nA^-1 (대각화 이용):")
sp.pprint(A_inv)

# 항등행렬 확인
I_check = A * A_inv
print("\nA * A^-1 결과 (항등행렬):")
sp.pprint(I_check)
```

```python
import sympy as sp

# 행렬 정의
A = sp.Matrix([[1, 2],
               [3, -4]])

# 대각화 수행
P, D = A.diagonalize()

print("원래 행렬 A:")
sp.pprint(A)

print("\n고유벡터 행렬 P:")
sp.pprint(P)

print("\n대각행렬 D:")
sp.pprint(D)

# 확인: A = P * D * P^-1
A_reconstructed = P * D * P.inv()
print("\nP * D * P^-1 결과:")
sp.pprint(A_reconstructed)

# 역행렬 확인: A^-1 = P * D^-1 * P^-1
D_inv = D.inv()
A_inv = P * D_inv * P.inv()
print("\nA^-1 (대각화 이용):")
sp.pprint(A_inv)

# 항등행렬 확인
I_check = A * A_inv
print("\nA * A^-1 결과 (항등행렬):")
sp.pprint(I_check)
```
