```python
from sympy import Matrix, symbols, solve, latex, eye,sqrt
from IPython.display import display, Math
import sympy as sp
λ,σ = symbols('λ σ') 
mat_A=Matrix([[1,1,0],[0,1,1]])
mat_A
```

#### 먼저 행렬 A를 정의합니다.
#### 그 다음 전치행렬을 정의합니다.

```python
mat_A_T=mat_A.T
mat_A_T
```

##### 두개의 행렬을 곱합니다.

```python
ATA=mat_A_T*mat_A
ATA
```

```python
AAT=mat_A*mat_A_T
AAT
```

```python
mat_I=eye(3)
```

##### A^T*A의 고유값을 구하면

```python
sik1=ATA-λ*mat_I
sik2=sik1.det()
roots=solve(sik2,λ)
roots=sorted(roots,reverse=True)
display(Math(" , ".join([latex(r) for r in roots])))
```

```python
singular_values=[sp.sqrt(r) for r in roots]
display(Math("σ: " + " , ".join([latex(s) for s in singular_values])))
```

구하여진 σ를 통해서 Σ를 구하면

```python
sigma=Matrix([[singular_values[0],0,0],[0,singular_values[1],0]])
sigma
```
