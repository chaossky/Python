```python
import sympy as sp

# 미지수 zeta 정의
zeta = sp.Symbol('zeta')

# 행렬 정의 (미지수 포함)
M = sp.Matrix([[0,2**(-0.5), 2**(-0.5)],
               [0,2**(-0.5), -2**(-0.5)],
               [zeta,0,0]])
M
```

```python
# 전치 행렬과 곱
ATA = M.T * M

ATA
```

```python
print("A^T A =")
sp.pprint(ATA)

# 직교 조건: A^T A = I
eq = sp.Eq(ATA, sp.eye(2))

# 방정식 풀기
solution = sp.solve(eq, zeta)
print("alpha solutions:", solution)
```

```python
import sympy as sp

# 1. 벡터 집합 정의
v1 = sp.Matrix([1, 1, 0])
v2 = sp.Matrix([1, 0, 1])
v3 = sp.Matrix([0, 1, 1])

vectors = [v1, v2, v3]

# 2. Gram-Schmidt 과정 직접 구현
orthogonal_vectors = []
orthonormal_vectors = []

for i, v in enumerate(vectors):
    # (a) 이전 벡터들의 성분 제거
    u = v
    for q in orthogonal_vectors:
        proj = (v.dot(q) / q.dot(q)) * q
        u = u - proj
    
    orthogonal_vectors.append(u)
    
    # (b) 정규화 (길이를 1로 맞춤)
    q_normalized = u / sp.sqrt(u.dot(u))
    orthonormal_vectors.append(q_normalized)

# 3. 결과 출력
print("원래 벡터 집합:")
for v in vectors:
    print(v)

print("\n직교 벡터 집합 (orthogonal):")
for u in orthogonal_vectors:
    print(u)

print("\n직교 정규 벡터 집합 (orthonormal):")
for q in orthonormal_vectors:
    print(q)
```

```python
import sympy as sp

def gram_schmidt(A):
    """
    Perform Gram-Schmidt orthogonalization on the columns of matrix A.
    Returns an orthonormal matrix Q (with Sympy exact arithmetic).
    """
    # Convert to sympy Matrix
    A = sp.Matrix(A)
    n = A.shape[1]
    Q = []
    
    for j in range(n):
        v = A[:, j]
        for i in range(j):
            proj = (Q[i].dot(v)) * Q[i]
            v = v - proj
        v = v / sp.sqrt(v.dot(v))  # normalize
        Q.append(v)
    
    return sp.Matrix.hstack(*Q)

# Define the matrix
A = sp.Matrix([[1, 3],
               [-3, 4]])

# Apply Gram-Schmidt
Q = gram_schmidt(A)

print("Original matrix A:")
sp.pprint(A)
print("\nOrthogonal (Orthonormal) matrix Q:")
sp.pprint(Q)

# Verify orthogonality: Q^T * Q should be identity
print("\nQ^T Q:")
sp.pprint(Q.T * Q)
```
