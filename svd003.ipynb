```python
import sympy as sp
from IPython.display import Math

# 행렬 정의
A = sp.Matrix([[1,1,0],[0,1,1]])
ATA = A.T * A
λ = sp.symbols('λ')

# 1. 특성방정식 계산
char_poly = (ATA - λ*sp.eye(ATA.shape[0])).det()
display(Math("특성방정식: " + sp.latex(char_poly)))

# 2. 고유값과 고유벡터를 함께 구하기
eig_pairs = []
for r in sp.solve(char_poly, λ):
    ns = (ATA - r*sp.eye(ATA.shape[0])).nullspace()
    for v in ns:
        eig_pairs.append((r, v.normalized()))

# 고유값 기준으로 내림차순 정렬
eig_pairs.sort(key=lambda x: x[0], reverse=True)

# 정렬된 순서대로 특이값과 벡터 추출
sigmas = [sp.sqrt(r) for r, v in eig_pairs]
V_cols = [v for r, v in eig_pairs]

display(Math("고유값 λ (σ^2): " + " , ".join([sp.latex(r) for r, v in eig_pairs])))
display(Math("특이값 σ: " + " , ".join([sp.latex(s) for s in sigmas])))

# V 행렬 구성
V = sp.Matrix.hstack(*V_cols)
display(Math("V = " + sp.latex(V)))

# 3. U 계산: u_i = (1/σ_i) A v_i
U_cols = []
for sigma, v in zip(sigmas, V_cols):
    if sigma != 0:
        u = (A*v)/sigma
        U_cols.append(u.normalized())

U = sp.Matrix.hstack(*U_cols)
display(Math("U = " + sp.latex(U)))

# 4. Σ 행렬 구성
Sigma = sp.zeros(A.shape[0], A.shape[1])
for i, sigma in enumerate(sigmas):
    if i < Sigma.shape[0]:
        Sigma[i,i] = sigma
display(Math("\\Sigma = " + sp.latex(Sigma)))

# 5. 검증: A ≈ U Σ V^T
reconstructed = U * Sigma * V.T
display(Math("U \\Sigma V^T = " + sp.latex(reconstructed)))
```

```python
import sympy as sp
from IPython.display import Math

# 행렬 정의
A = sp.Matrix([[1,1,0],[0,1,1]])
ATA = A.T * A
λ = sp.symbols('λ')

# 고유값과 고유벡터 구하기
eig_pairs = []
for r in sp.solve((ATA - λ*sp.eye(ATA.shape[0])).det(), λ):
    ns = (ATA - r*sp.eye(ATA.shape[0])).nullspace()
    for v in ns:
        eig_pairs.append((r, v.normalized()))

# 내림차순 정렬
eig_pairs.sort(key=lambda x: x[0], reverse=True)

# 특이값과 V 추출
sigmas = [sp.sqrt(r) for r, v in eig_pairs]
V_cols = [v for r, v in eig_pairs]
V = sp.Matrix.hstack(*V_cols)

# Σ 행렬 구성
Sigma = sp.zeros(A.shape[0], A.shape[1])
for i, sigma in enumerate(sigmas):
    if i < Sigma.shape[0]:
        Sigma[i,i] = sigma

# A*V 계산
AV = A * V
display(Math("A V = " + sp.latex(AV)))

# 각 열을 대응하는 σ로 나누어 U 계산
U_cols = []
for sigma, col in zip(sigmas, AV.T):
    if sigma != 0:
        u = (col.T / sigma).T  # σ로 나누기
        U_cols.append(u.normalized())
U = sp.Matrix.hstack(*U_cols)

display(Math("U = " + sp.latex(U)))
display(Math("\\Sigma = " + sp.latex(Sigma)))
display(Math("V = " + sp.latex(V)))

# 검증: U Σ V^T
reconstructed = U * Sigma * V.T
display(Math("U \\Sigma V^T = " + sp.latex(reconstructed)))
```

```python
# A*V 계산
AV = A * V
display(Math("A V = " + sp.latex(AV)))

# 각 열을 대응하는 σ로 나누어 U 계산
U_cols = []
for sigma, col in zip(sigmas, AV.cols):  # AV.cols로 열벡터를 바로 가져옴
    if sigma != 0:
        u = col / sigma                  # σ로 나누기
        U_cols.append(u.normalized())

U = sp.Matrix.hstack(*U_cols)

display(Math("U = " + sp.latex(U)))
display(Math("\\Sigma = " + sp.latex(Sigma)))
display(Math("V = " + sp.latex(V)))

# 검증: U Σ V^T
reconstructed = U * Sigma * V.T
display(Math("U \\Sigma V^T = " + sp.latex(reconstructed)))
```
