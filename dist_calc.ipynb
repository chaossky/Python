```python
from sympy import symbols, sqrt, Rational, latex
from IPython.display import display, Math

def distance2D(x1,y1,x2,y2):
    expr=sqrt((x2-x1)**2+(y2-y1)**2)
    display(Math(latex(expr)))
    
def distance2D_point(pt1,pt2):
    x1,y1=pt1
    x2,y2=pt2
    expr=sqrt((x2-x1)**2+(y2-y1)**2)
    display(Math(latex(expr)))



def point_to_line_distance_latex_render(point, line, form="general"):
    """
    점과 직선 사이의 최단 거리를 LaTeX 형태로 렌더링해서 출력하는 함수
    """
    p1, q1 = point
    
    if form == "general":
        a, b, c = line
    elif form == "slope_intercept":
        m, n = line
        a, b, c = m, -1, n
    else:
        raise ValueError("form은 'general' 또는 'slope_intercept'만 가능합니다.")
    
    numerator = abs(a * p1 + b * q1 + c)
    denominator = sqrt(a**2 + b**2)
    
    expr = Rational(numerator, 1) / denominator
    latex_expr = latex(expr)
    
    # Jupyter Notebook에서 LaTeX 수식 렌더링
    display(Math(latex_expr))

# 사용 예시
point_to_line_distance_latex_render((0, 0), (2, 3), form="slope_intercept")
```

```python
from sympy import symbols, sqrt, Rational, latex

def point_to_line_distance_latex(point, line, form="general"):
    """
    점과 직선 사이의 최단 거리를 LaTeX 형태로 반환하는 함수
    """
    p1, q1 = point
    
    if form == "general":
        a, b, c = line
    elif form == "slope_intercept":
        m, n = line
        a, b, c = m, -1, n
    else:
        raise ValueError("form은 'general' 또는 'slope_intercept'만 가능합니다.")
    
    numerator = abs(a * p1 + b * q1 + c)
    denominator = sqrt(a**2 + b**2)
    display(Math(latex(numerator)))
    
    display(Math(latex(denominator)))
    
    expr = Rational(numerator, 1) / denominator
    display(Math(latex(expr)))

# 사용 예시
point_to_line_distance_latex((0, 0), (1, -1, -3), form="general")
# 출력: \sqrt{2}

point_to_line_distance_latex((0, 0), (2, 3), form="slope_intercept")
# 출력: \frac{3}{\sqrt{5}}
```

```python
point_to_line_distance_latex((2, 3), (5, 12, -7), form="general")
```

```python
point_to_line_distance_latex((3, -5), (2, -1, -1), form="general")
```

```python
point_to_line_distance_latex((-4, 1), (3, -4, 6), form="general")
```
