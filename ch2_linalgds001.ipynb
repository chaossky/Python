```python
import sympy as sp 
def least_squares(A, b): 
    """ 행렬 A와 벡터 b에 대해 최소제곱 해 x를 구하는 함수 
    x = (A^T A)^(-1) A^T b 
    """ 
    # 정규방정식 계산 
    x = (A.T * A).inv() * A.T * b 
    return x 

# 예시 사용 
A = sp.Matrix([[1, 2], [3, 1], [1, 1]]) 
b = sp.Matrix([[5.5], [-3.5], [1.5]]) 
x = least_squares(A, b) 
print("최소제곱 해 x =", x)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 벡터 정의
w = np.array([3, 2])
v = np.array([5, 1])

# projection (수평 성분)
wp = (np.dot(w, v) / np.dot(v, v)) * v

# orthogonal component (수직 성분)
wt = w - wp

plt.figure(figsize=(10,6))
plt.axhline(0, color='gray', linewidth=0.5)
plt.axvline(0, color='gray', linewidth=0.5)

# 원래 벡터 w
plt.quiver(0, 0, w[0], w[1], angles='xy', scale_units='xy', scale=1, color='blue', label='w')

# 벡터 v
plt.quiver(0, 0, v[0], v[1], angles='xy', scale_units='xy', scale=1, color='green', label='v')

# w의 평행 성분 wp (점선으로 표시)
plt.plot([0, wp[0]], [0, wp[1]], 'r--', label='w_p (parallel)')

# w의 수직 성분 wt (wp 끝점에서 시작, 점선으로 표시)
plt.plot([wp[0], w[0]], [wp[1], w[1]], 'orange', linestyle='--', label='w_t (perpendicular)')

# 보기 좋게 설정
plt.xlim(-1, 6)
plt.ylim(-1, 4)
plt.gca().set_aspect('equal', adjustable='box')
plt.legend()
plt.title("벡터 w의 분해: 평행 성분 w_p와 수직 성분 w_t")
plt.show()
```

```python
import numpy as np
import plotly.graph_objects as go
import plotly.io as pio 
pio.renderers.default = "browser"

# 벡터 정의
w = np.array([3, 2, 4])
v = np.array([2, 1, 1])

# projection (평행 성분)
wp = (np.dot(w, v) / np.dot(v, v)) * v

# orthogonal component (수직 성분)
wt = w - wp

# 3D 그래프 생성
fig = go.Figure()

# 원래 벡터 w
fig.add_trace(go.Scatter3d(
    x=[0, w[0]], y=[0, w[1]], z=[0, w[2]],
    mode='lines+markers',
    line=dict(color='blue', width=6),
    marker=dict(size=4),
    name='w'
))

# 벡터 v
fig.add_trace(go.Scatter3d(
    x=[0, v[0]], y=[0, v[1]], z=[0, v[2]],
    mode='lines+markers',
    line=dict(color='green', width=6),
    marker=dict(size=4),
    name='v'
))

# 평행 성분 wp
fig.add_trace(go.Scatter3d(
    x=[0, wp[0]], y=[0, wp[1]], z=[0, wp[2]],
    mode='lines+markers',
    line=dict(color='red', width=4, dash='dash'),
    marker=dict(size=4),
    name='w_p (parallel)'
))

# 수직 성분 wt (wp에서 w까지)
fig.add_trace(go.Scatter3d(
    x=[wp[0], w[0]], y=[wp[1], w[1]], z=[wp[2], w[2]],
    mode='lines+markers',
    line=dict(color='orange', width=4, dash='dash'),
    marker=dict(size=4),
    name='w_t (perpendicular)'
))

# 레이아웃 설정
fig.update_layout(
    scene=dict(
        xaxis=dict(range=[0,5], title='X'),
        yaxis=dict(range=[0,5], title='Y'),
        zaxis=dict(range=[0,5], title='Z'),
        aspectmode='cube'
    ),
    title="R^3 공간에서 벡터 w의 분해",
    legend=dict(x=0.02, y=0.95)
)

fig.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data
n = 4
x = np.linspace(-1, 1, n)
y = np.linspace(-1, 1, n)
z = np.linspace(-1, 1, n)
X, Y, Z = np.meshgrid(x, y, z)
U = (X + Y)/5
V = (Y - X)/5
W = Z*0


# Plot
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.quiver(X, Y, Z, U, V, W)

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# ===== 평면 벡터 방정식 입력 =====
# r0: 평면 위의 한 점
r0 = np.array([1, 2, 3])

# u, v: 평면을 span하는 두 방향 벡터
u = np.array([2, -1, 1])
v = np.array([1, 1, 2])

# ===== 평면 생성 =====
s_vals = np.linspace(-5, 5, 20)
t_vals = np.linspace(-5, 5, 20)
S, T = np.meshgrid(s_vals, t_vals)

X = r0[0] + S * u[0] + T * v[0]
Y = r0[1] + S * u[1] + T * v[1]
Z = r0[2] + S * u[2] + T * v[2]

# ===== 시각화 =====
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# 평면
ax.plot_surface(X, Y, Z, alpha=0.5, color='skyblue', edgecolor='gray')

# 기준점 표시
ax.scatter(r0[0], r0[1], r0[2], color='red', s=50, label='r0')

# 축 설정
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.legend()
plt.show()
```
