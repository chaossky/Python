```python
import numpy as np

# 행렬 A
A = np.array([[1, 2],
              [3, -4]])

# 고유벡터 행렬 V
V = np.array([[2, 1],
              [1, -3]])

# 고유값 대각행렬 D
D = np.diag([2, -5])

# D의 역행렬
D_inv = np.diag([1/2, -1/5])

# A의 역행렬 (V D^-1 V^-1)
A_inv = V @ D_inv @ np.linalg.inv(V)

print("A inverse:\n", A_inv)

# 확인: A * A^-1 = I
I_check = A @ A_inv
print("A * A^-1:\n", I_check)

 
```
