```python
from sympy import Matrix, latex
from IPython.display import display, Math

# 예시 행렬 (중복된 고유값을 가지도록 설정)
mat = Matrix([[3, 0, 0],
              [-2, 4, 0],
              [-4,2, 3]])

# 고유값 딕셔너리
eigen_data = mat.eigenvals()

# 중복도만큼 고유값을 반복해서 리스트에 추가
eigen_list = []
for eigenvalue, multiplicity in eigen_data.items():
    eigen_list.extend([latex(eigenvalue)] * multiplicity)

# 쉼표로 연결
eigen_str = ", ".join(eigen_list)

# 출력
display(Math(r"\lambda = " + eigen_str))
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

# 예시 행렬
A = Matrix([[3, 1],
            [4, 6]])

# 고유값 + 고유벡터 구하기
eigen_data = A.eigenvects()

# LaTeX으로 출력
for eigenvalue, multiplicity, eigenvectors in eigen_data:
    display(Math(r"\lambda = " + latex(eigenvalue)))
    for vec in eigenvectors:
        display(Math(r"\text{고유벡터} = " + latex(vec)))
```

```python
from sympy import Matrix, latex
from IPython.display import display, Math

# 예시 행렬
A = Matrix([[3, 1],
            [4, 6]])

# 고유값과 고유벡터 구하기
eigen_data = A.eigenvects()

# enumerate를 이용해 첨자 붙이기
for idx, (eigenvalue, multiplicity, eigenvectors) in enumerate(eigen_data, start=1):
    display(Math(r"\lambda_{" + str(idx) + "} = " + latex(eigenvalue)))
    for vec in eigenvectors:
        display(Math(r"\vec {v}_{ "+str(idx) +"} ="+  latex(vec)))
```

```python
import numpy as np 
A = np.array([[4, 1], [0, 2]]) 
w, v = np.linalg.eig(A) 
D = np.diag(w) 
P = v # 확인: A ≈ P @ D @ np.linalg.inv(P)
print(P)
```
