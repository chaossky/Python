```python
import sympy as sp
from IPython.display import display, Latex ,Math
```

```python
# 랜덤 정수 행렬 생성 함수
def random_matrix(rows, cols):
    return sp.Matrix(sp.randMatrix(rows, cols, min=-5, max=5))
```

```python
# 세 가지 형태의 행렬 생성
A_square = random_matrix(3, 3)
A_wide   = random_matrix(3, 5)
A_tall   = random_matrix(5, 3)

# 행렬 출력
display(Math(r"\textbf{Random Matrices:}"))
display(Math(r"A_{\text{square}} = " + sp.latex(A_square)))
display(Math(r"A_{\text{wide}} = " + sp.latex(A_wide)))
display(Math(r"A_{\text{tall}} = " + sp.latex(A_tall)))
```

```python
# SVD 비교 함수
def show_svd_info(A, name):
    display(Math(r"\textbf{Full SVD for }" + name))
    U, S, V = A.singular_value_decomposition()
    display(Math(r"U: " + str(U.shape) + r", \; S: " + str(S.shape) + r", \; V: " + str(V.shape)))
    
    # Economy SVD 직접 구현
    r = min(A.shape)
    Ue = U[:, :r]
    Se = S[:r, :r]
    Ve = V[:r, :]
    
    display(Math(r"\textbf{Economy SVD for }" + name))
    display(Math(r"U_e: " + str(Ue.shape) + r", \; S_e: " + str(Se.shape) + r", \; V_e: " + str(Ve.shape)))
```

```python
# 실행
show_svd_info(A_square, "A_{square}")
show_svd_info(A_wide, "A_{wide}")
show_svd_info(A_tall, "A_{tall}")
```
