```python
# Sympy 불러오기
import sympy as sp

# 1. 행렬 정의
A = sp.Matrix([[1, 1, 0],
               [0, 1, 1]])
print("행렬 A:")
sp.pprint(A)

# 2. A^T * A 계산
ATA = A.T * A
print("\nA^T * A:")
sp.pprint(ATA)

# 3. 고유값과 고유벡터 구하기
eigs = ATA.eigenvects()
print("\nA^T A의 고유값과 고유벡터:")
for val, mult, vecs in eigs:
    print("고유값:", val)
    print("고유벡터:", vecs)

# 4. 특이값 (고유값의 제곱근)
singular_values = [sp.sqrt(val) for val, _, _ in eigs]
print("\n특이값 Σ:")
sp.pprint(singular_values)

# 5. 오른쪽 특이벡터 V (정규화)
V = sp.Matrix.hstack(*[vec[0].normalized() for _, _, vecs in eigs for vec in vecs])
print("\n오른쪽 특이벡터 행렬 V:")
sp.pprint(V)

# 6. 왼쪽 특이벡터 U 계산 (u_i = (1/σ_i) * A v_i)
U_cols = []
for i, sigma in enumerate(singular_values):
    if sigma != 0:
        u = (A * V[:, i]) / sigma
        U_cols.append(u.normalized())
U = sp.Matrix.hstack(*U_cols)
print("\n왼쪽 특이벡터 행렬 U:")
sp.pprint(U)

# 7. Σ 행렬 구성
Sigma = sp.zeros(A.shape[0], A.shape[1])
for i, sigma in enumerate(singular_values):
    if i < Sigma.shape[0]:
        Sigma[i, i] = sigma
print("\nΣ 행렬:")
sp.pprint(Sigma)

# 8. 검증: A == U * Σ * V.T
print("\n검증 (U Σ V^T):")
sp.pprint(U * Sigma * V.T)
```

```python
import sympy as sp

# 행렬 정의
A = sp.Matrix([[1,1,0],[0,1,1]])
ATA = A.T * A
λ = sp.symbols('λ')

# 1. 특성방정식 (determinant)
char_poly = ATA.det()  # 잘못된 방식: 전체 행렬식
# 올바른 방식은 det(ATA - λI)
char_poly = (ATA - λ*sp.eye(ATA.shape[0])).det()
print("특성방정식:", char_poly)

# 2. 고유값 구하기
roots = sp.solve(char_poly, λ)
print("고유값:", roots)

# 3. 각 고유값에 대해 고유벡터 구하기
eigenvectors = []
for r in roots:
    # (ATA - λI) v = 0 의 해 공간(nullspace)을 구함
    null_space = (ATA - r*sp.eye(ATA.shape[0])).nullspace()
    eigenvectors.append((r, null_space))

# 4. 출력
for val, vecs in eigenvectors:
    print(f"\n고유값 λ = {val}")
    for v in vecs:
        print("고유벡터:", v.normalized())
```

```python
import sympy as sp

A = sp.Matrix([[1,1,0],[0,1,1]])
ATA = A.T * A
λ = sp.symbols('λ')

# 1. 특성방정식
char_poly = (ATA - λ*sp.eye(ATA.shape[0])).det()
roots = sp.solve(char_poly, λ)  # 고유값 λ
print("고유값 λ:", roots)

# 2. 특이값 σ = sqrt(λ)
sigmas = [sp.sqrt(r) for r in roots]
print("특이값 σ:", sigmas)

# 3. 각 λ에 대해 고유벡터 v 구하기
V_cols = []
for r in roots:
    null_space = (ATA - r*sp.eye(ATA.shape[0])).nullspace()
    for v in null_space:
        V_cols.append(v.normalized())

V = sp.Matrix.hstack(*V_cols)
print("오른쪽 특이벡터 V:")
sp.pprint(V)

# 4. U 계산: u_i = (1/σ_i) A v_i
U_cols = []
for sigma, v in zip(sigmas, V_cols):
    if sigma != 0:
        u = (A*v)/sigma
        U_cols.append(u.normalized())

U = sp.Matrix.hstack(*U_cols)
print("왼쪽 특이벡터 U:")
sp.pprint(U)

# 5. Σ 행렬
Sigma = sp.zeros(A.shape[0], A.shape[1])
for i, sigma in enumerate(sigmas):
    if i < Sigma.shape[0]:
        Sigma[i,i] = sigma
print("Σ 행렬:")
sp.pprint(Sigma)
```

```python
import sympy as sp
from IPython.display import Math

# 행렬 정의
A = sp.Matrix([[1,1,0],[0,1,1]])
ATA = A.T * A
AAT=A*A.T
λ = sp.symbols('λ')

# 1. 특성방정식 계산
char_poly = (ATA - λ*sp.eye(ATA.shape[0])).det()
display(Math("특성방정식: " + sp.latex(char_poly)))

# 2. 고유값 구하기 (σ^2 값)
roots = sp.solve(char_poly, λ)
roots=sorted(roots,reverse=True)
display(Math("고유값 λ (σ^2): " + " , ".join([sp.latex(r) for r in roots])))

# 3. 특이값 σ = sqrt(λ)
sigmas = [sp.sqrt(r) for r in roots]
display(Math("특이값 σ: " + " , ".join([sp.latex(s) for s in sigmas])))

# 4. 각 λ에 대해 고유벡터 v 구하기
V_cols = []
for r in roots:
    ns = (ATA - r*sp.eye(ATA.shape[0])).nullspace()
    for v in ns:
        V_cols.append(v.normalized())

# V 행렬 구성
V = sp.Matrix.hstack(*V_cols)
display(Math("V = " + sp.latex(V)))

# 5. U 계산: u_i = (1/σ_i) A v_i
U_cols = []
for sigma, v in zip(sigmas, V_cols):
    if sigma != 0:
        u = (A*v)/sigma
        U_cols.append(u.normalized())

U = sp.Matrix.hstack(*U_cols)
display(Math("U = " + sp.latex(U)))

# 6. Σ 행렬 구성
Sigma = sp.zeros(A.shape[0], A.shape[1])
for i, sigma in enumerate(sigmas):
    if i < Sigma.shape[0]:
        Sigma[i,i] = sigma
display(Math("\\Sigma = " + sp.latex(Sigma)))

# 7. 검증: A ≈ U Σ V^T
reconstructed = U * Sigma * V.T
display(Math("U \\Sigma V^T = " + sp.latex(reconstructed)))
```
