int, list, dict 도 사실 모두 클래스
이 클래스를 인스턴스를 만들고 매소드를 사용

```python
a=int(10)
a
```

```python
b=list(range(10))
b
```

```python
c=dict(x=10,y=10)
c
```

```python
b.append(20)
b
```
